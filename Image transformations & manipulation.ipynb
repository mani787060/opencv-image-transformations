```python
import cv2 
```

```python
# loading/reading the image :-

image = cv2.imread("cat pic.jpg")
```

```python
# Resizing & Rescaling images :-

if image is None:
    print("Image is not found")

else:
    print("Image loaded")

    resized = cv2.resize(image, (400, 400))   # note it is written in the form of (width, height)

    cv2.imshow("Original Image", image)
    cv2.imshow("Resized Image", resized)

    # note that you have to saved your 'resized_image' in the last 
    # otherwise, our output image will be overlap.
    cv2.imwrite("resized_output.jpg", resized)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Cropping the images :-

if image is not None:

    cropped = image[50:200, 50:150]

    cv2.imshow("Original", image) 
    cv2.imshow("Cropped image", cropped) 

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Rotating an image :-

if image is None:
    print("could not load image")

else:
    (h,w) = image.shape[:2]     # we use shape[:2] b/c we want 2 values

    # for finding center
    center = (w//2, h//2)
       
    # we give (angle_value = 90) for clockwise rotation and (scale = 1) that means we want image of same size.
    M = cv2.getRotationMatrix2D(center, -65, 1.0)   # can give (angle = -90) for getting image in 'anticlockwise_direction'
    
    
    rotated = cv2.warpAffine(image, M, (w,h))


    cv2.imshow("Original", image)
    cv2.imshow("Rotated 90 degree", rotated)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# flipping an image :-

if image is None:
    print("could not load an image")

else:
    flipped_horizontal = cv2.flip(image, 1)
    flipped_vertical = cv2.flip(image, 0)
    flipped_both = cv2.flip(image, -1)

    # displaying each image one by one 
    cv2.imshow("Original", image)
    cv2.imshow("Flipped horizontal", flipped_horizontal)
    cv2.imshow("Flipped vertical", flipped_vertical)
    cv2.imshow("Flipped both", flipped_both)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
